```python
import json\nfrom pathlib import Path\nimport matplotlib.pyplot as plt\n\nmetrics_path = Path('results_unsloth_colab/unsloth_metrics.json')\nwith metrics_path.open() as f:\n    m = json.load(f)\n\nbase = m['baseline']['mean_reward']\nft = m['finetuned']['mean_reward']\nprint('Baseline mean reward:', round(base, 2))\nprint('Finetuned mean reward:', round(ft, 2))\nprint('Improvement:', round(m['reward_improvement'], 2))\n\nplt.figure(figsize=(5,3.5))\nplt.bar(['baseline','finetuned'], [base, ft], color=['#97a5bf','#46b96a'])\nplt.ylabel('Mean reward')\nplt.title('Varaha Unsloth Reward Improvement')\nplt.grid(axis='y', alpha=0.2)\nplt.show()
```
